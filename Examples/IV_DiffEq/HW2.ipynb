```python
import numpy as np
import matplotlib.pyplot as plt

# Time grid
t = np.linspace(0, 2*np.pi, 800)

# Solutions from Method (2)
x = 2*np.sin(2*t) + 3*np.cos(t)
y = 2*np.cos(2*t) - np.sin(t)

# Key points
points = {
    r"$\left(0,\frac{1}{2}\right)$": (0, 0.5),
    r"$(0,-3)$": (0, -3),
    r"$(0,-1)$": (0, -1),
}

# Create figure
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(x, y, label=r"Trajectory $(x(t),y(t))$ for $0 \leq t<2\pi$", lw=2)

# Mark and label special points
for label, (xp, yp) in points.items():
    ax.scatter(xp, yp, color='crimson', s=60, zorder=3)
    ax.text(xp + 0.15, yp, label, fontsize=11, va='center')
ax.scatter(3, 2, s=60, zorder=3)
ax.text(3 + 0.25, 2, r"$t=0$", fontsize=11, va='center')

# Axes, style
ax.axhline(0, color='black', lw=0.8)
ax.axvline(0, color='black', lw=0.8)
ax.set_aspect('equal', adjustable='box')
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$y$")
ax.set_title("Phase-plane trajectory for the forced linear system")
ax.legend(frameon=False)
ax.grid(True, alpha=0.3)
ax.axis('equal')

plt.tight_layout() 
plt.show()
```
